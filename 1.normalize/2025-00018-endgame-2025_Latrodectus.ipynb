## Read in files, guess column and type

```python
files=sorted(glob(f"{IN_DIR}/*.tsv"))
```

```python
files
```

```python
goodlines = 0
badlines = 0
lines = []
for file in files:
    filename = os.path.basename(file)
    df = pd.read_csv(file, delimiter='\t')
    for index, row in df.iterrows():
        extra_data = {
            "last_used" : row["last_used"],
            "masked_password" : row["password"]
        }
        fields = [ row["username"], row["password"], row["url"] ]
        fields.extend(defaults)
        fields[-1] = json.dumps(extra_data)
        lines.append(fields)
    column_names= ["username","passwd","url", "ts_found", "ts_leaked", "has_name", "has_dob", "has_addr", "has_phone", "has_cc", "has_bankacc", "has_ssn", 
                         "has_ip", "extra_data" ] 
    
    out_df = pd.DataFrame(lines, columns=column_names)

    out_df.to_csv(f"{OUT_DIR}/{filename.replace('.txt','.tsv')}",sep="\t", index=False, 
                  columns=column_names
                 )
```
